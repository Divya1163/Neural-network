```python
# YOLOv8 + OpenCV (Beginner, simple cells)
import os
import cv2
from ultralytics import YOLO
```

```python
# Mode settings
use_webcam = False
image_path = "image.jpeg"
confidence_threshold = 0.45
```

```python
# Load YOLOv8 model (downloads once)
model = YOLO("yolov8n.pt")
print("YOLOv8 model loaded")
print("Class count:", len(model.names))
```

```python
# Image detection block
if not use_webcam and not os.path.exists(image_path):
    print("Image not found. Update image_path.")
    print("Current path:", image_path)
```

```python
if not use_webcam and os.path.exists(image_path):
    results = model(image_path, conf=confidence_threshold)
    annotated = results[0].plot()
    print("Detected objects:", len(results[0].boxes))
```

```python
if not use_webcam and os.path.exists(image_path):
    cv2.imshow("YOLOv8 Image Detection", annotated)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Webcam detection block
if use_webcam:
    cap = cv2.VideoCapture(0)
    print("Press q to quit webcam detection")
```

```python
if use_webcam and cap.isOpened():
    while True:
        ok, frame = cap.read()
        if not ok:
            break

        results = model(frame, conf=confidence_threshold, verbose=False)
        view = results[0].plot()
        cv2.imshow("YOLOv8 Webcam Detection", view)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
```

```python
print("\nNotes:")
print("1) YOLOv8 gives modern object detection with class labels.")
print("2) Lower confidence_threshold = more detections, more false positives.")
print("3) Set use_webcam=True for live detection.")
```
