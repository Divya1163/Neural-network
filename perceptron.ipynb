# AND Gate Perceptron
Pure basic perceptron implementation for AND gate (no libraries, no custom functions).

```python
# Perceptron from scratch (no libraries, no custom functions)

# Training data for AND gate
x1 = [0, 0, 1, 1]
x2 = [0, 1, 0, 1]
y  = [0, 0, 0, 1]

# Initial weights and bias
w1 = 0
w2 = 0
b = 0

# Learning rate
lr = 1

# Train for multiple passes (epochs)
epoch = 0
while epoch < 10:
    i = 0
    total_error = 0

    while i < 4:
        # Weighted sum
        z = x1[i] * w1 + x2[i] * w2 + b

        # Step activation
        if z >= 0:
            pred = 1
        else:
            pred = 0

        # Perceptron update rule
        error = y[i] - pred
        w1 = w1 + lr * error * x1[i]
        w2 = w2 + lr * error * x2[i]
        b = b + lr * error

        if error < 0:
            total_error = total_error - error
        else:
            total_error = total_error + error

        i = i + 1

    print("Epoch", epoch + 1, "| w1 =", w1, "w2 =", w2, "b =", b, "| total error =", total_error)

    if total_error == 0:
        break

    epoch = epoch + 1

print("\nFinal model:")
print("w1 =", w1, "w2 =", w2, "b =", b)

# Test on all inputs
print("\nPredictions:")
j = 0
while j < 4:
    z = x1[j] * w1 + x2[j] * w2 + b
    if z >= 0:
        out = 1
    else:
        out = 0
    print(x1[j], x2[j], "->", out)
    j = j + 1
```

# Decision Boundary Plots
Input points and perceptron decision line for each gate.

```python
import matplotlib.pyplot as plt

gates = ["AND", "OR", "NAND"]
params = [
    [1, 1, -1],
    [1, 1, -1],
    [-1, -1, 1]
 ]

truth = [
    [0, 0, 0, 1],
    [0, 1, 1, 1],
    [1, 1, 1, 0]
 ]

x_data = [0, 0, 1, 1]
y_data = [0, 1, 0, 1]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

k = 0
while k < 3:
    ax = axes[k]
    gate = gates[k]
    w1 = params[k][0]
    w2 = params[k][1]
    b = params[k][2]
    t = truth[k]

    pos_x = []
    pos_y = []
    neg_x = []
    neg_y = []

    i = 0
    while i < 4:
        if t[i] == 1:
            pos_x.append(x_data[i])
            pos_y.append(y_data[i])
        else:
            neg_x.append(x_data[i])
            neg_y.append(y_data[i])
        i = i + 1

    ax.scatter(neg_x, neg_y, s=120, c="#d62728", label="Class 0")
    ax.scatter(pos_x, pos_y, s=120, c="#2ca02c", label="Class 1")

    x_line = [-0.5, 1.5]
    if w2 != 0:
        y0 = (-w1 * x_line[0] - b) / w2
        y1 = (-w1 * x_line[1] - b) / w2
        ax.plot(x_line, [y0, y1], c="#1f77b4", linewidth=2, label="Decision line")

    ax.set_title(gate + " Decision Boundary")
    ax.set_xlim(-0.5, 1.5)
    ax.set_ylim(-0.5, 1.5)
    ax.set_xlabel("x1")
    ax.set_ylabel("x2")
    ax.grid(alpha=0.3)
    ax.legend(loc="upper right", fontsize=8)

    k = k + 1

plt.tight_layout()
plt.show()
```

# Result Comparison Visuals
Truth-table bars and confusion-style matrices for AND, OR, and NAND.

```python
import matplotlib.pyplot as plt

gates = ["AND", "OR", "NAND"]
params = [
    [1, 1, -1],
    [1, 1, -1],
    [-1, -1, 1]
 ]

inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
 ]

truth = [
    [0, 0, 0, 1],
    [0, 1, 1, 1],
    [1, 1, 1, 0]
 ]

# Compute predictions for each gate
all_preds = []
k = 0
while k < 3:
    w1 = params[k][0]
    w2 = params[k][1]
    b = params[k][2]
    preds = []

    i = 0
    while i < 4:
        z = inputs[i][0] * w1 + inputs[i][1] * w2 + b
        if z >= 0:
            preds.append(1)
        else:
            preds.append(0)
        i = i + 1

    all_preds.append(preds)
    k = k + 1

# Plot 1: truth vs prediction bars
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

k = 0
while k < 3:
    ax = axes[k]
    x_pos = [0, 1, 2, 3]
    width = 0.35

    truth_vals = truth[k]
    pred_vals = all_preds[k]

    truth_left = [p - width / 2 for p in x_pos]
    pred_right = [p + width / 2 for p in x_pos]

    ax.bar(truth_left, truth_vals, width=width, label="Truth", color="#4c78a8")
    ax.bar(pred_right, pred_vals, width=width, label="Pred", color="#f58518")

    ax.set_title(gates[k] + " Truth vs Pred")
    ax.set_xticks(x_pos)
    ax.set_xticklabels(["00", "01", "10", "11"])
    ax.set_ylim(0, 1.2)
    ax.set_xlabel("Input")
    ax.set_ylabel("Output")
    ax.legend(fontsize=8)
    ax.grid(axis="y", alpha=0.25)

    k = k + 1

plt.tight_layout()
plt.show()

# Plot 2: confusion-style matrix for each gate
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

k = 0
while k < 3:
    t = truth[k]
    p = all_preds[k]

    tn = 0
    fp = 0
    fn = 0
    tp = 0

    i = 0
    while i < 4:
        if t[i] == 0 and p[i] == 0:
            tn = tn + 1
        elif t[i] == 0 and p[i] == 1:
            fp = fp + 1
        elif t[i] == 1 and p[i] == 0:
            fn = fn + 1
        else:
            tp = tp + 1
        i = i + 1

    matrix = [[tn, fp], [fn, tp]]

    ax = axes[k]
    im = ax.imshow(matrix, cmap="Blues", vmin=0, vmax=4)
    ax.set_title(gates[k] + " Confusion Matrix")
    ax.set_xticks([0, 1])
    ax.set_yticks([0, 1])
    ax.set_xticklabels(["Pred 0", "Pred 1"])
    ax.set_yticklabels(["True 0", "True 1"])

    r = 0
    while r < 2:
        c = 0
        while c < 2:
            ax.text(c, r, str(matrix[r][c]), ha="center", va="center", color="black", fontsize=11)
            c = c + 1
        r = r + 1

    k = k + 1

plt.tight_layout()
plt.show()
```
